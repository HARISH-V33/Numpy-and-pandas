## 1. What is Pandas?

Pandas is a Python library for working with **labelled, table-like data** (rows and columns), like an Excel sheet or a SQL table, but inside Python. It is built on top of NumPy. It has two main objects:

- **Series**: one labelled column (1D)
- **DataFrame**: a full table of many columns (2D)

**Analogy:** A NumPy array is a plain grid of numbers with no names. Pandas is the same grid with **row labels and column names**, so you can say "give me the salary of Raj" instead of "give me row 3, column 2".

## 2. Why Pandas?

| | NumPy array | Pandas DataFrame |
|---|---|---|
| Data types | One type for the whole array | **Different type for each column** |
| Labels | Only positions (0, 1, 2) | Row index and column names |
| Missing values | Only NaN in float arrays | Built-in missing value support |
| File reading | Limited | `read_csv`, `read_excel`, `read_json`, `read_sql` |
| Group, merge, pivot | Not available | Built in |
| Best for | Pure numeric math | Real-world mixed datasets |

**Real-world example:** A hospital keeps patient records (name, age, city, blood group, visit date). Different types in different columns, so it fits a DataFrame, not a NumPy array.

**Business example:** A sales team has an Excel file of 50,000 orders. With Pandas they load it, filter the orders of one city, and find the monthly total in a few lines.

**AI/ML example:** Almost every ML project starts with `df = pd.read_csv(...)`. We inspect, clean, create features and then convert `df` into NumPy arrays for scikit-learn or deep learning models.

**Topics covered:** What and Why Pandas, Series, DataFrame, Creating DataFrames, Data Types, Index, Columns, Reading Data, Writing Data.

## 3. Series

A **Series** is a **1D labelled array**. It has two parts:
- **values**: the data (a NumPy array inside)
- **index**: the labels for each value

**Analogy:** One column of an Excel sheet, with the row names on the left.

| Create from | Example |
|---|---|
| List | `pd.Series([80, 90, 75])` |
| List with index | `pd.Series([80, 90, 75], index=["Raj", "Anu", "Kumar"])` |
| Dictionary | `pd.Series({"Raj": 80, "Anu": 90})` (keys become the index) |
| Scalar | `pd.Series(5, index=["a", "b", "c"])` |

**Key features:**
- Access by label (`s["Anu"]`) or by position (`s.iloc[1]`)
- Math is vectorized (`s * 2`, `s > 80`)
- **Automatic alignment:** when two Series are added, Pandas matches them by **index label**, not by position. Labels present in only one Series give `NaN`.

**Business example:** Monthly sales of one product, with the month names as the index.

**AI/ML example:** A single feature column (`df["age"]`) or the target column (`df["label"]`) is a Series. Model predictions are often returned as a Series or array.

```python
marks = pd.Series([80, 90, 75], index=["Raj", "Anu", "Kumar"], name="marks")
print(marks)
print("\nValues:", marks.values)
print("Index :", marks.index)
print("dtype :", marks.dtype)

# Access
print("\nmarks['Anu']  :", marks["Anu"])
print("marks.iloc[0] :", marks.iloc[0])
print("First two:\n", marks.iloc[:2])

# Vectorized operations and filtering
print("\nmarks + 5:\n", marks + 5)
print("Marks above 78:\n", marks[marks > 78])
print("Mean:", marks.mean(), "| Max:", marks.max(), "| Name of topper:", marks.idxmax())

# From a dictionary
sales = pd.Series({"Jan": 100, "Feb": 120, "Mar": 90})
print("\nFrom dict:\n", sales)

# Alignment by index label
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s2 = pd.Series([10, 20, 30], index=["b", "c", "d"])
print("\ns1 + s2 (aligned by label):\n", s1 + s2)
print("\nWith fill_value=0:\n", s1.add(s2, fill_value=0))
```

**Output Explanation:**
- The Series shows the index on the left and the values on the right, and at the bottom `Name: marks, dtype: int64`.
- `marks["Anu"]` uses the **label**, and `marks.iloc[0]` uses the **position**. `idxmax()` returns the **label** of the largest value (Anu), not the value.
- `marks + 5` adds 5 to every value, and `marks[marks > 78]` keeps only Raj and Anu (Boolean filtering, as in NumPy).
- In `s1 + s2`, only `b` and `c` exist in both, so they are added (22 and 33). `a` and `d` exist in only one Series, so the result is `NaN`. Using `.add(..., fill_value=0)` treats the missing side as 0.
- Because of this alignment, the result becomes **float** (NaN is a float).

## 4. DataFrame

A **DataFrame** is a **2D labelled table**. Each column is a Series, and all columns share the same row index. Every column can have its own data type.

**Analogy:** An Excel sheet, or a SQL table. A DataFrame is a **dictionary of Series**.

| Attribute | Meaning |
|---|---|
| `df.shape` | (rows, columns) |
| `df.columns` | Column names |
| `df.index` | Row labels |
| `df.dtypes` | Data type of each column |
| `df.values` / `df.to_numpy()` | The data as a NumPy array |
| `df.size`, `df.ndim` | Total cells, number of dimensions (2) |

**Business example:** A customer table with `customer_id`, `name`, `city`, `monthly_spend`, `joined_on`.

**AI/ML example:** The **dataset**: each row is an observation (sample), and each column is a feature, and one column is the target. We split it into `X = df[features]` and `y = df["target"]`.

## 5. Creating DataFrames

| From | Example |
|---|---|
| Dictionary of lists | `pd.DataFrame({"name": [...], "age": [...]})` (keys become column names) |
| List of dictionaries | `pd.DataFrame([{"name": "Raj", "age": 20}, ...])` (each dict is a row) |
| List of lists | `pd.DataFrame(rows, columns=[...])` |
| NumPy array | `pd.DataFrame(arr, columns=[...])` |
| Series | `pd.DataFrame({"a": s1, "b": s2})` |
| File | `pd.read_csv(...)` (covered in the Reading Data section) |

**Real-world example:** API responses usually come as a **list of dictionaries** (JSON), which become a DataFrame with one line.

**AI/ML example:** Predictions and features from NumPy arrays are wrapped into a DataFrame for analysis: `pd.DataFrame(X_scaled, columns=feature_names)`.

```python
# 1. Dictionary of lists
df = pd.DataFrame({
    "name": ["Raj", "Anu", "Kumar", "Divya"],
    "age": [20, 22, 21, 22],
    "city": ["Madurai", "Chennai", "Trichy", "Chennai"],
    "marks": [80.5, 90.0, 75.5, 88.0],
    "passed": [True, True, True, True],
})
print(df)

# 2. List of dictionaries (like an API response)
records = [
    {"name": "Meena", "age": 23, "city": "Salem"},
    {"name": "Arun", "age": 24},                      # 'city' missing -> NaN
]
print("\nFrom list of dicts:\n", pd.DataFrame(records))

# 3. List of lists with column names
rows = [["Raj", 80], ["Anu", 90]]
print("\nFrom list of lists:\n", pd.DataFrame(rows, columns=["name", "marks"]))

# 4. NumPy array
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("\nFrom NumPy array:\n", pd.DataFrame(arr, columns=["a", "b", "c"]))

# 5. From Series
s1 = pd.Series([1, 2, 3], index=["x", "y", "z"])
s2 = pd.Series([10, 20, 30], index=["x", "y", "z"])
print("\nFrom Series:\n", pd.DataFrame({"first": s1, "second": s2}))

# Attributes
print("\nshape  :", df.shape)
print("columns:", list(df.columns))
print("index  :", df.index)
print("size   :", df.size, "| ndim:", df.ndim)
print("Back to NumPy:\n", df[["age", "marks"]].to_numpy())
```

**Output Explanation:**
- A dictionary of lists gives **one column per key**. All lists must have the same length, otherwise Pandas raises a `ValueError`.
- A list of dictionaries gives **one row per dictionary**. A key missing in a dictionary (`city` for Arun) becomes `NaN`, so Pandas handles ragged data nicely.
- A NumPy array has no names, so we pass `columns=[...]`.
- `shape (4, 5)` means 4 rows and 5 columns, `size` is 4 x 5 = 20 cells, and `ndim` is always 2.
- `to_numpy()` converts the DataFrame back to a NumPy array, which is what scikit-learn uses.

## 6. Data Types (dtypes)

Each column has **one data type**. The type decides what operations are possible, how much memory is used, and how fast it runs.

| dtype | Meaning | Example |
|---|---|---|
| `int64` | Whole numbers | age, count |
| `float64` | Decimal numbers (also used when NaN is present) | marks, price |
| `bool` | True / False | passed |
| `object` / `str` | Text (older versions show `object`, newer versions may show `str`) | name, city |
| `datetime64[ns]` | Dates and times | order_date |
| `category` | A limited set of repeated labels (saves memory) | gender, city |
| `Int64`, `boolean` | Nullable types that allow missing values without becoming float | |

**Converting types:**
- `df["col"].astype(int)` for a simple conversion
- `pd.to_numeric(df["col"], errors="coerce")` converts text to numbers, bad values become `NaN`
- `pd.to_datetime(df["col"])` converts text to dates

**Business example:** A price column stored as text (`"1,200"`) cannot be summed until it is cleaned and converted to a number.

**AI/ML example:** ML models need **numeric** input. We check `dtypes`, convert numbers stored as text, convert dates into parts, and encode categories into numbers.

```python
print(df.dtypes)

# Messy data: numbers stored as text
raw = pd.DataFrame({
    "price": ["100", "250", "abc", "80"],
    "date": ["2025-01-05", "2025-02-10", "2025-03-15", "2025-04-20"],
    "grade": ["A", "B", "A", "C"],
})
print("\nBefore:\n", raw.dtypes)

raw["price"] = pd.to_numeric(raw["price"], errors="coerce")     # 'abc' becomes NaN
raw["date"] = pd.to_datetime(raw["date"])
raw["grade"] = raw["grade"].astype("category")
print("\nAfter:\n", raw.dtypes)
print("\n", raw)

# astype for simple conversions
print("\nmarks as int:", df["marks"].astype(int).tolist(), "(decimal part is cut)")

# Why dtype matters: text cannot be summed numerically
text_prices = pd.Series(["100", "200"])
print("\nText 'sum':", text_prices.sum(), "(joins the text, not 300)")
print("Numeric sum:", pd.to_numeric(text_prices).sum())

# Select columns by dtype
print("\nNumeric columns:", list(df.select_dtypes(include="number").columns))
```

**Output Explanation:**
- The first DataFrame shows `int64` for age, `float64` for marks, `bool` for passed, and `object` (or `str` in newer Pandas versions) for text columns.
- `to_numeric(errors="coerce")` converted the bad value `"abc"` into `NaN` instead of crashing, and the price column became `float64`.
- `to_datetime` created a `datetime64[ns]` column, so we can extract month, year and so on (Notebook 11).
- `category` stores repeated labels efficiently (Notebook 14).
- Summing text **joins** the strings (`"100200"`) and does not add them, which is why the dtype must be correct before calculating.
- `select_dtypes` picks the numeric columns, which is a common step before scaling.

## 7. Index

The **index** holds the **row labels**. By default it is `0, 1, 2, ...` (a `RangeIndex`), but it can be anything: names, IDs, dates.

| Operation | Code | Meaning |
|---|---|---|
| Set a column as index | `df.set_index("name")` | Column becomes the row labels |
| Move the index back to a column | `df.reset_index()` | Index becomes a normal column |
| Drop the old index | `df.reset_index(drop=True)` | Gives a clean `0, 1, 2...` |
| Access by label | `df.loc["Raj"]` | Row with label "Raj" |
| Access by position | `df.iloc[0]` | First row |
| Rename the index | `df.index.name = "student"` | |

**Properties:**
- Index labels need not be unique (but unique is better).
- The index is **immutable** (single labels cannot be changed, we replace the whole index).
- Pandas **aligns data by index** in operations (as we saw in Series).
- After filtering or sorting, the old index labels stay, so use `reset_index(drop=True)` for a clean one.

**Business example:** A product table with `product_id` as the index makes lookups by ID quick and clear.

**AI/ML example:** A **time series** uses dates as the index. After a train-test split, the DataFrames keep the original row labels, which helps to trace a prediction back to its source row.

```python
print("Default index:", df.index)

indexed = df.set_index("name")
print("\nset_index('name'):\n", indexed)
print("\nindexed.loc['Anu']:\n", indexed.loc["Anu"])
print("\nindexed.iloc[0]:\n", indexed.iloc[0])

# Back to a normal column
print("\nreset_index():\n", indexed.reset_index().head(2))

# After filtering, the old labels stay
adults = df[df["age"] >= 22]
print("\nFiltered (old index stays):\n", adults)
print("\nAfter reset_index(drop=True):\n", adults.reset_index(drop=True))

# Dates as an index
dates = pd.date_range("2025-01-01", periods=4, freq="D")
daily = pd.Series([100, 110, 95, 120], index=dates)
print("\nDate index:\n", daily)
print("Sales on 2025-01-03:", daily.loc["2025-01-03"])

# Index is immutable
try:
    df.index[0] = 99
except TypeError as e:
    print("\nError:", e)

# Rename the index and the columns axis
indexed.index.name = "student"
print("\nIndex name:", indexed.index.name)
```

**Output Explanation:**
- `set_index("name")` moves the names to the left as row labels, so `loc["Anu"]` finds Anu's row directly. `iloc[0]` still works by position.
- `reset_index()` brings the labels back as a normal column.
- The filtered DataFrame **keeps the original labels** (1, 2, 3 and so on), not 0, 1, 2. That is why `loc` and `iloc` can give different answers on filtered data. `reset_index(drop=True)` creates a clean new index.
- A date index lets us select by date (`loc["2025-01-03"]`), and it is the base of time series analysis.
- Changing a single index label fails with a `TypeError` because the index is immutable.

## 8. Columns

Columns are the **variables (features)** of the table. We can select, add, rename and delete them.

| Task | Code |
|---|---|
| One column (Series) | `df["age"]` |
| Many columns (DataFrame) | `df[["name", "age"]]` (double brackets) |
| List all columns | `df.columns`, `list(df.columns)` |
| Add or change a column | `df["pass_mark"] = df["marks"] >= 80` |
| Rename | `df.rename(columns={"marks": "score"})` |
| Delete | `df.drop(columns=["passed"])` |
| Reorder | `df[["city", "name", "age"]]` |

**Dot access (`df.age`) vs bracket access (`df["age"]`):** The bracket way is **always safe**. The dot way fails if the column name has a space, or if it clashes with a DataFrame method (for example a column named `count`), and it **cannot create** new columns.

**Business example:** Adding a `total = price * quantity` column to an orders table.

**AI/ML example:** **Feature engineering**: new columns are created from old ones (`bmi = weight / height ** 2`). Then `X = df[feature_columns]` selects the model inputs.

*(Detailed renaming, dropping and `assign()` are covered in Notebook 8.)*

```python
print("Columns:", list(df.columns))

print("\ndf['age'] is a", type(df["age"]).__name__)
print("df[['name', 'age']] is a", type(df[["name", "age"]]).__name__)
print(df[["name", "age"]])

# Add new columns
df["bonus_marks"] = df["marks"] + 5
df["grade"] = np.where(df["marks"] >= 85, "A", "B")
print("\nAfter adding columns:\n", df)

# Rename and drop (return new DataFrames, original unchanged)
renamed = df.rename(columns={"marks": "score"})
dropped = df.drop(columns=["bonus_marks", "passed"])
print("\nRenamed columns:", list(renamed.columns))
print("Dropped columns:", list(dropped.columns))
print("Original still has:", list(df.columns))

# ML style: features and target
features = ["age", "marks"]
X = df[features]
y = df["grade"]
print("\nX shape:", X.shape, "| y shape:", y.shape)

# Dot access trap
tricky = pd.DataFrame({"count": [1, 2], "total sales": [10, 20]})
print("\ntricky.count is a method:", callable(tricky.count))
print("Safe access:", tricky["count"].tolist(), tricky["total sales"].tolist())
```

**Output Explanation:**
- `df["age"]` returns a **Series**, and `df[["name", "age"]]` (double brackets, a list inside) returns a **DataFrame**.
- Assigning to a new column name creates the column (`bonus_marks`, `grade`). `np.where` works as a vectorized if-else.
- `rename` and `drop` returned **new** DataFrames, and the original `df` still has all its columns. (Pandas does not change the original unless you reassign it or use `inplace=True`, which is not recommended.)
- `X` has 2 columns and `y` is a single column with the same number of rows: the standard ML split.
- `tricky.count` is the **method** `DataFrame.count`, not our column, so we must use `tricky["count"]`.

## 9. Reading Data and Writing Data

Pandas reads many formats into a DataFrame, and writes a DataFrame back to many formats.

| Format | Read | Write |
|---|---|---|
| CSV | `pd.read_csv("file.csv")` | `df.to_csv("file.csv", index=False)` |
| Excel | `pd.read_excel("file.xlsx")` (needs `openpyxl`) | `df.to_excel("file.xlsx", index=False)` |
| JSON | `pd.read_json("file.json")` | `df.to_json("file.json", orient="records")` |
| Parquet | `pd.read_parquet("file.parquet")` | `df.to_parquet("file.parquet")` |
| SQL | `pd.read_sql(query, connection)` | `df.to_sql("table", connection)` |

**Useful `read_csv` options:**

| Option | Use |
|---|---|
| `usecols=["a", "b"]` | Read only some columns (saves memory) |
| `nrows=100` | Read only the first 100 rows (quick look at a huge file) |
| `dtype={"id": str}` | Force column types |
| `parse_dates=["date"]` | Convert columns to dates while reading |
| `index_col="id"` | Use a column as the index |
| `sep=";"` | If the file uses another separator |
| `na_values=["NA", "?"]` | Extra symbols to treat as missing |

**Always write with `index=False`**, otherwise the index is saved as an extra column and appears as `Unnamed: 0` when you read the file again.

**Business example:** Reading the daily sales CSV exported from the billing software, and saving the cleaned report as Excel for the manager.

**AI/ML example:** Reading the training data (`read_csv`), saving the cleaned dataset as Parquet (fast and small), and saving predictions to CSV for submission.

*(Excel, Parquet, Pickle, SQL and multiple files are covered in detail in Notebook 5.)*

```python
# Write: save the DataFrame as CSV (without the index)
df.to_csv("students.csv", index=False)

# Read it back
loaded = pd.read_csv("students.csv")
print(loaded)
print("\nSame shape?", loaded.shape == df.shape)

# The index mistake: write WITH the index
df.to_csv("students_with_index.csv")
bad = pd.read_csv("students_with_index.csv")
print("\nColumns after saving with the index:", list(bad.columns)[:3])

# read_csv options
subset = pd.read_csv("students.csv", usecols=["name", "marks"], nrows=2)
print("\nusecols + nrows:\n", subset)

typed = pd.read_csv("students.csv", index_col="name")
print("\nindex_col='name':\n", typed.head(2))

# JSON
df.to_json("students.json", orient="records", indent=2)
from_json = pd.read_json("students.json")
print("\nFrom JSON:\n", from_json.head(2))

# Excel (needs: pip install openpyxl)
try:
    df.to_excel("students.xlsx", index=False)
    print("\nExcel rows read back:", len(pd.read_excel("students.xlsx")))
except ImportError:
    print("\nopenpyxl not installed. Run: pip install openpyxl")

# Missing file
try:
    pd.read_csv("no_such_file.csv")
except FileNotFoundError as e:
    print("\nError:", e)

# Quick look at the data
print("\n", loaded.head(2))
print(loaded.tail(2))
print(loaded.info())
```

**Output Explanation:**
- `to_csv(..., index=False)` saves only the data columns. Reading it back gives the same shape.
- When we saved **with** the index, reading the file produced an extra column named `Unnamed: 0`. This is the most common beginner mistake in file handling.
- `usecols` and `nrows` read only part of the file, which is very useful for huge files.
- `index_col="name"` makes the `name` column the index while reading.
- JSON saved with `orient="records"` becomes a list of dictionaries, one per row, the same shape as an API response.
- A missing file raises `FileNotFoundError` (the same exception from Python Sprint 1), which we can handle with `try/except`.
- `head()`, `tail()` and `info()` give a quick check right after loading. They are covered fully in Notebook 6.

## Best Practices

- Start with `df.head()`, `df.shape`, `df.info()` and `df.dtypes` after loading any data.
- Use **bracket access** `df["col"]`, not dot access `df.col`.
- Use **double brackets** `df[["a", "b"]]` for several columns.
- Always save with **`index=False`** unless the index has meaning.
- Convert types early (`to_numeric`, `to_datetime`, `astype("category")`).
- Give clear, consistent column names (lowercase, no spaces, such as `monthly_spend`).
- Use `.copy()` when you create a new DataFrame from a slice that you will modify.
- Read only the columns you need (`usecols`) for large files.
- Keep the **raw file untouched** and save cleaned data to a new file.

## Common Mistakes

| Mistake | Problem | Fix |
|---|---|---|
| Saving with the index | Extra `Unnamed: 0` column on reload | `to_csv(..., index=False)` |
| `df["a", "b"]` | `KeyError` | Use `df[["a", "b"]]` |
| Using `df.count` or `df.size` as a column name | It returns a method or attribute | Use `df["count"]` |
| Numbers stored as text | Wrong sums, sort order and joins | `pd.to_numeric(..., errors="coerce")` |
| Expecting `drop` or `rename` to change the original | Original unchanged | Assign the result: `df = df.drop(...)` |
| Dictionary of lists with different lengths | `ValueError` | Make all lists equal in length |
| Mixing up `loc` and `iloc` on filtered data | Wrong rows selected | `loc` = label, `iloc` = position |
| Not resetting the index after filtering | Index gaps cause confusion | `reset_index(drop=True)` |
| Reading a huge file completely | Memory error | Use `usecols`, `nrows` or `chunksize` |

## Advantages
- Labelled data, so code is readable (`df["salary"]` instead of `arr[:, 3]`)
- Different data type for each column, so real mixed tables fit naturally
- Easy reading and writing of CSV, Excel, JSON, SQL and Parquet
- Rich tools for cleaning, filtering, grouping, merging and time series
- Works smoothly with NumPy, Matplotlib and scikit-learn

## Limitations
- Slower and heavier in memory than NumPy for pure numeric math
- The whole DataFrame is loaded in memory, so very large data needs chunking, Dask or Spark
- Mainly single-core by default
- Many ways to do the same thing can confuse beginners (`loc` vs `iloc` vs `[]`)
- Behaviour and defaults can change between versions (for example, text dtype and copy rules in newer releases)
